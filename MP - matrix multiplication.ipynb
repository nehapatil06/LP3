```python
# Matrix Multiplication

r1 = int(input("Enter rows of Matrix A: "))
c1 = int(input("Enter columns of Matrix A: "))

r2 = int(input("Enter rows of Matrix B: "))
c2 = int(input("Enter columns of Matrix B: "))

if c1 != r2:
    print("Matrix multiplication not possible")

else:
    print("Enter Matrix A:")
    A = []
    for i in range(r1):
        A.append(list(map(int, input().split())))

    print("Enter Matrix B:")
    B = []
    for i in range(r2):
        B.append(list(map(int, input().split())))

    C = [[0 for j in range(c2)] for i in range(r1)]

    for i in range(r1):
        for j in range(c2):
            for k in range(c1):
                C[i][j] += A[i][k] * B[k][j]

    print("Result:")
    for row in C:
        print(row)
```

```python
# Multithreaded Matrix Multiplication (One Thread Per Row)

import threading

def multiply_row(i):
    for j in range(c2):
        for k in range(c1):
            C[i][j] += A[i][k] * B[k][j]

r1 = int(input("Enter rows of Matrix A: "))
c1 = int(input("Enter columns of Matrix A: "))

r2 = int(input("Enter rows of Matrix B: "))
c2 = int(input("Enter columns of Matrix B: "))

if c1 != r2:
    print("Multiplication not possible")

else:
    print("Enter Matrix A:")
    A = [list(map(int, input().split())) for i in range(r1)]

    print("Enter Matrix B:")
    B = [list(map(int, input().split())) for i in range(r2)]

    C = [[0 for j in range(c2)] for i in range(r1)]

    threads = []

    for i in range(r1):
        t = threading.Thread(target=multiply_row, args=(i,))
        threads.append(t)
        t.start()

    for t in threads:
        t.join()

    print("Result:")
    for row in C:
        print(row)
```
