```python
# N-Queens using Backtracking

def is_safe(board, row, col, n):

    for i in range(row):
        if board[i][col] == 1:
            return False

    i = row - 1
    j = col - 1
    while i >= 0 and j >= 0:
        if board[i][j] == 1:
            return False
        i -= 1
        j -= 1

    i = row - 1
    j = col + 1
    while i >= 0 and j < n:
        if board[i][j] == 1:
            return False
        i -= 1
        j += 1

    return True


def solve(board, row, n):

    if row == n:
        return True

    for col in range(n):
        if is_safe(board, row, col, n):
            board[row][col] = 1

            if solve(board, row + 1, n):
                return True

            board[row][col] = 0

    return False


# Main program
n = int(input("Enter the value of N: "))

board = [[0 for j in range(n)] for i in range(n)]

first_col = int(input("Enter column for first Queen (0 to n-1): "))

board[0][first_col] = 1

if solve(board, 1, n):
    print("\nFinal N-Queens Matrix:")
    for row in board:
        print(row)
else:
    print("No solution exists.")
```
